```python
from getpass import getpass

# Refresh key/client to avoid using stale or invalid values (e.g., "sk-")
deepseek_api_key = os.getenv("DEEPSEEK_API_KEY") or os.getenv("API_TOKEN")
if (
    not deepseek_api_key
    or deepseek_api_key.strip() == "sk-"
    or len(deepseek_api_key.strip()) < 20
    or "ollama-local" in deepseek_api_key.lower()
):
    print("DeepSeek key is missing/invalid in environment. Please enter a valid key.")

    entered_key = getpass("DEEPSEEK_API_KEY: ").strip()
    if (
        not entered_key
        or entered_key == "sk-"
        or len(entered_key) < 20
        or "ollama-local" in entered_key.lower()
    ):
        raise ValueError(
            "Invalid DeepSeek API key. Set a valid DEEPSEEK_API_KEY (or API_TOKEN) in .env, then reload the notebook."
        )

    deepseek_api_key = entered_key
    os.environ["DEEPSEEK_API_KEY"] = deepseek_api_key

openai_python_client = OpenAI(
    api_key=deepseek_api_key,
    base_url=deepseek_base_url
)

response = openai_python_client.chat.completions.create(
    model="deepseek-chat",
    messages=messages,
    stream=False
)
raw_response = response.choices[0].message.content
display(Markdown(f"**Response from Deepseek API:** {raw_response}"))
```

```python
# Ollama Python
from ollama import chat

def ai_agent(user_input):
    response = chat(
        model="llama3",
        messages=[{"role": "system", 
                   "content":"you are an intelligent AI Assistance helpful assistant that"
                   " provides concise and accurate answers to user queries."},
                  {"role": "user", "content": user_input}]
    )
    return response["message"]["content"]

while True:
    question = input("Enter your question (or type 'exit' to quit): ")
    if question.lower() == 'exit':
        break
    answer = ai_agent(question)
    print(f"AI Agent Response: {answer}")
```

```python
from ollama import chat

def calculator(expression):
    return eval(expression)

while True:
    query = input("Enter your question or 'calculate <expression>' (type 'exit' to quit): ")

    if query.lower() == "exit":
        print("Goodbye!")
        break

    # Calculator
    if query.lower().startswith("calculate"):
        expression = query.replace("calculate", "", 1).strip()

        try:
            result = calculator(expression)
            print(f"Result: {result}")
        except Exception as e:
            print(f"Calculation Error: {e}")

    # AI Chat
    else:
        try:
            response = chat(
                model="llama3",
                messages=[
                    {
                        "role": "system",
                        "content": (
                            "You are an intelligent AI assistant that provides "
                            "concise and accurate answers to user queries."
                        ),
                    },
                    {
                        "role": "user",
                        "content": query,
                    },
                ],
            )

            print(f"\nAI: {response.message.content}\n")

        except Exception as e:
            print(f"Ollama Error: {e}")
```
